```python
import pandas as pd
import matplotlib.pyplot as plt

username = "user"
password = "123"
attempts = 3
logged_in = False

while attempts > 0:
    username_input = input("Enter your username: ")
    password_input = input("Enter your password: ")
    
    if username_input == username and password_input == password:
        print("\nLogin successful!\n")
        logged_in = True
        break
    else:
        attempts -= 1
        if attempts > 0:
            print("Incorrect credentials. Attempts left:",attempts,"\n")
        else:
            print("No attempts left. Login failed.\n")

if logged_in:
    while True:
        print("\nMAIN MENU")
        print("1. Upload Data")
        print("2. Data Analysis")
        print("3. Data Manipulation")
        print("4. Data Visualization")
        print("5. Export Data")
        print("6. Exit")

        try:
            choice = int(input("Enter a choice (1-6): "))
        except ValueError:
            print("Invalid input. Please enter a number between 1 and 6.")
            continue

        if choice == 1:
            data = input("Enter the CSV file path to upload: ")
            try:
                df = pd.read_csv(data)

                print("\nDo you want to set the first column of the data as the index?")
                print("Y - Yes")
                print("N - No")
                set_index_choice = input("Choose (Y/N): ")

                if set_index_choice.lower() == 'y':
                    df = df.set_index(df.columns[0])
                    print("First column set as index.")
                else:
                    print("Index left unchanged.")

                print("Data uploaded successfully.")

            except Exception as e:
                print("Error loading data:", e)

        elif choice == 2:
            try:
                df
            except NameError:
                print("No dataset found. Please upload data first.")
                continue

            while True:
                print("\nDATA ANALYSIS MENU")
                print("1. Top 5 Rows")
                print("2. Bottom 5 Rows")
                print("3. Specific Data View")
                print("4. Data Info")
                print("5. Describe Data")
                print("6. View Columns")
                print("7. Sort Data")
                print("8. Exit")

                try:
                    choice_DA = int(input("Enter a choice (1-8): "))
                except ValueError:
                    print("Invalid input. Please enter a number between 1 and 8.")
                    continue

                if choice_DA == 1:
                    print(df.head())
                elif choice_DA == 2:
                    print(df.tail())
                elif choice_DA == 3:
                    print("L - View data using labels")
                    print("I - View data using indices")
                    loc_or_iloc = input("Choose (L/I): ")
                    try:
                        if loc_or_iloc.lower() == "l":
                            col_loc_start = input("Enter the first column label: ")
                            col_loc_end = input("Enter the last column label: ")
                            row_loc_start = input("Enter the first row label: ")
                            row_loc_end = input("Enter the last row label: ")
                            print(df.loc[row_loc_start:row_loc_end,col_loc_start:col_loc_end])
                        elif loc_or_iloc.lower() == "i":
                            col_iloc_start = int(input("Enter the first column index: "))
                            col_iloc_end = int(input("Enter the last column index: "))
                            row_iloc_start = int(input("Enter the first row index: "))
                            row_iloc_end = int(input("Enter the last row index: "))
                            print(df.iloc[row_iloc_start:row_iloc_end+1,col_iloc_start:col_iloc_end+1])
                        else:
                            print("Invalid choice. Please enter 'L' or 'I'.")
                    except Exception as e:
                        print("Error:", e)
                elif choice_DA == 4:
                    print(df.info())
                elif choice_DA == 5:
                    print(df.describe())
                elif choice_DA == 6:
                    print(df.columns)
                elif choice_DA == 7:
                    sort_column = input("Enter the column to sort by: ")
                    if sort_column not in df.columns:
                        print("Invalid column name.")
                    else:
                        print("A - Ascending")
                        print("D - Descending")
                        sort_choice = input("Choose sorting order (A/D): ")
                        ascending = sort_choice.lower() == "a"
                        print(df.sort_values(by=sort_column, ascending=ascending))
                elif choice_DA == 8:
                    break
                else:
                    print("Invalid choice. Please enter a number between 1 and 8.")

        elif choice == 3:
            try:
                df
            except NameError:
                print("No dataset found. Please upload data first.")
                continue

            while True:
                print("\nDATA MANIPULATION MENU")
                print("1. Delete a Row")
                print("2. Delete a Column")
                print("3. Exit")

                try:
                    choice_DM = int(input("Enter a choice (1-3): "))
                except ValueError:
                    print("Invalid input. Please enter a number between 1 and 3.")
                    continue

                if choice_DM == 1:
                    index = int(input("Enter the row index to delete: "))
                    if index not in df.index:
                        print("Invalid row index.")
                    else:
                        print("T - Delete Temporarily")
                        print("P - Delete Permanently")
                        per_row = input("Choose (T/P): ")
                        if per_row.lower() == "t":
                            print(df.drop(index, axis=0))
                        elif per_row.lower() == "p":
                            df.drop(index, axis=0, inplace=True)
                            print("Row deleted permanently.")
                        else:
                            print("Invalid choice.")
                elif choice_DM == 2:
                    col_del = input("Enter the column name to delete: ")
                    if col_del not in df.columns:
                        print("Invalid column name.")
                    else:
                        print("T - Delete Temporarily")
                        print("P - Delete Permanently")
                        per_col = input("Choose (T/P): ")
                        if per_col.lower() == "t":
                            print(df.drop(col_del, axis=1))
                        elif per_col.lower() == "p":
                            df.drop(col_del, axis=1, inplace=True)
                            print("Column deleted permanently.")
                        else:
                            print("Invalid choice.")
                elif choice_DM == 3:
                    break
                else:
                    print("Invalid choice. Please enter a number between 1 and 3.")

        elif choice == 4:
            try:
                df
            except NameError:
                print("No dataset found. Please upload data first.")
                continue

            while True:
                print("\nDATA VISUALIZATION MENU")
                print("1. Bar Graph")
                print("2. Line Chart")
                print("3. Histogram")
                print("4. Exit")

                try:
                    choice_DV = int(input("Enter a choice (1-4): "))
                except ValueError:
                    print("Invalid input. Please enter a number between 1 and 4.")
                    continue

                if choice_DV == 4:
                    break
                elif choice_DV not in [1, 2, 3]:
                    print("Invalid choice. Please enter a valid number.")
                    continue

                x = input("Enter the column for X-axis: ")
                if x not in df.columns:
                    print("Invalid X-axis column.")
                    continue

                y = None
                if choice_DV in [1, 2]:
                    y = input("Enter the column for Y-axis: ")
                    if y not in df.columns:
                        print("Invalid Y-axis column.")
                        continue

                xlabel = input("Enter X-axis label: ")
                ylabel = input("Enter Y-axis label: ")
                title = input("Enter chart title: ")

                try:
                    df_clean = df.dropna(subset=[x, y])

                    grouped = df_clean.groupby(x)[y].mean()

                    if choice_DV == 1: 
                        print("V - Vertical Bar Graph")
                        print("H - Horizontal Bar Graph")
                        bar_ch = input("Choose (V/H): ")

                        if bar_ch.lower() == "v":
                            grouped.plot(kind='bar')
                        elif bar_ch.lower() == "h":
                            grouped.plot(kind='barh')
                        else:
                            print("Invalid choice.")
                            continue

                    elif choice_DV == 2:  
                        grouped.plot(kind='line', marker='o')

                    elif choice_DV == 3:   
                        plt.hist(df[x].dropna(), bins=5)

                    plt.xlabel(xlabel)
                    plt.ylabel(ylabel)
                    plt.title(title)
                    plt.tight_layout()
                    plt.show()

                except Exception as e:
                    print("Error generating chart:", e)

        elif choice == 5:
            try:
                df.to_csv("output.csv", index=True)
                print("Data exported successfully to output.csv")
            except Exception as e:
                print("Error exporting data:", e)

        elif choice == 6:
            print("Exiting program.")
            break

        else:
            print("Invalid choice. Please enter a number between 1 and 6.")
```
